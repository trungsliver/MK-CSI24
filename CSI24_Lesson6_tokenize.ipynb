## 1. Dữ liệu mẫu (Sentences)
- Dùng để huấn luyện Tokenizer
- Đây là dữ liệu đầu vào mà mô hình sẽ chuyển thành chuỗi số

```python
# Dữ liệu mẫu
sentences = [
    "Tôi thích học khoá Computer Science.",
    "Bạn có đam mê khoa học máy tính không?",
    "Huy thích lập trình giải quyết vấn đề.",
    "An muốn lập trình game bằng Python.",
    "Mỹ có nhiều dự án Scratch hay.",
    "Em thấy lập trình giống với trò chơi giải đố.",
    "Không có gì đã bằng code chạy lần đầu tiên đúng luôn",
    "Sở thích của mình là lập trình các dự án hay"
]
print(sentences)
```

## 2. Tạo và huấn luyện Tokenizer

```python
# Import thư viện
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
# Khởi tạo Tokenizer
    # num_words = 100: Chỉ dữ lại 100 từ phổ biến nhất trong tập dữ liệu
    # oov_token="<OOV>": token đặc biệt danh cho các từ không nằm trong từ điển (100 từ trong tập dữ liệu)
tokenizer = Tokenizer(num_words = 100, oov_token="<OOV>")
```

```python
# Huấn luyện Tokenizer trên dữ liệu
    # fit_on_texts(): phân tích các câu, xây dựng từ điển (word_index) dựa trên tần suất xuất hiện của các từ.
tokenizer.fit_on_texts(sentences)
```

```python
# In ra từ điển
from pprint import pprint
word_index = tokenizer.word_index
pprint(word_index)
```

## 3. Chuyển đổi văn bản thành chuỗi số (Sequences)

`tokenizer.texts_to_sequences(sentences)` sẽ chuyển đổi mỗi câu trong danh sách `sentences` thành một chuỗi các số nguyên, nơi mỗi số đại diện cho một từ theo từ điển đã học. Ví dụ, nếu từ 'thích' có chỉ số là 4, thì nó sẽ được thay bằng số 4 trong chuỗi.

```python
test_data = [
    'Lâm Khánh thích game.',
    'Linh thích Computer Science'
]
test_seq = tokenizer.texts_to_sequences(test_data)

# Các từ "Lâm", "Khánh", "Linh" không có trong từ điển => được gán bằng oov_token (1)
for i in range(len(test_data)):
  print(test_data[i])
  print(test_seq[i])
```

```python
# Test 3 câu đầu tiên của danh sách sentences
sequences = tokenizer.texts_to_sequences(sentences)
for i in range(3):
  print(sentences[i])
  print(sequences[i])
```

## 4. Padding chuỗi số

Để đưa các chuỗi số về cùng một độ dài, chúng ta sử dụng `pad_sequences` từ `tensorflow.keras.utils`.

*   `maxlen=10`: Đặt độ dài tối đa của mỗi chuỗi là 10.
*   `truncating="post"`: Nếu chuỗi dài hơn `maxlen`, nó sẽ bị cắt từ phía sau.
*   `padding="post"`: Nếu chuỗi ngắn hơn `maxlen`, nó sẽ được thêm số 0 vào phía sau cho đủ độ dài.

```python
from tensorflow.keras.utils import pad_sequences

padded = pad_sequences(test_seq, maxlen=10, truncating="post", padding="post")
print(padded)
```
